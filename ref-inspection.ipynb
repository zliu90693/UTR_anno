```python
import pandas as pd

Lzep_txinfo = pd.read_csv("./metadata/Lzep_txinfo.tsv", sep="\t")
Lzep_txinfo
```

```python
utr5_missing = ((Lzep_txinfo["utr5_len"] == 0) & 
                (Lzep_txinfo["cds_len"] != 0)).sum() / \
               (Lzep_txinfo["cds_len"] != 0).sum()
print(utr5_missing)
```

```python
utr3_missing = ((Lzep_txinfo["utr3_len"] == 0) & 
                (Lzep_txinfo["cds_len"] != 0)).sum() / \
               (Lzep_txinfo["cds_len"] != 0).sum()
print(utr3_missing)
```

```python
both_missing = ((Lzep_txinfo["utr3_len"] == 0) & 
                (Lzep_txinfo["utr5_len"] == 0) &
                (Lzep_txinfo["cds_len"] != 0)).sum() / \
               (Lzep_txinfo["cds_len"] != 0).sum()
print(both_missing)
```

```python
either_missing = (((Lzep_txinfo["utr3_len"] == 0) | (Lzep_txinfo["utr5_len"] == 0)) & # 注意!! 必须将 | 括起来, 因为 & 的优先级比 | 大
                (Lzep_txinfo["cds_len"] != 0)).sum() / \
               (Lzep_txinfo["cds_len"] != 0).sum()
print(either_missing)
```

The proportion of missing UTRs in *L. zephyrus* is excessively high, necessitating manual annotation.

Manually annotate UTRs in the GTF file using the [UTR_anno project](https://github.com/zliu90693/UTR_anno)

The GTF file containing manually annotated UTRs is named Lasioglossum_zephyrus_utranno.gtf.
