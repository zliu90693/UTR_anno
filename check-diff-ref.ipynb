```python
from pyfaidx import Fasta
import pandas as pd

fa = Fasta("./Jones_NEE_2023_Lzep/ref/Lasioglossum_zephyrus.fasta")
gtf_chroms = set()
with open("./Jones_NEE_2023_Lzep/ref/Lasioglossum_zephyrus.filtered.gtf") as f:
    for line in f:
        if not line.startswith("#"):
            gtf_chroms.add(line.split("\t")[0])

annotated_len = sum(len(fa[s]) for s in fa.keys() if s in gtf_chroms)
unannotated_len = sum(len(fa[s]) for s in fa.keys() if s not in gtf_chroms)
total_len = annotated_len + unannotated_len

print(f"已注释总长:{annotated_len/1e6:.1f} Mb ({annotated_len/total_len*100:.1f}%)")
print(f"未注释总长:{unannotated_len/1e6:.1f} Mb ({unannotated_len/total_len*100:.1f}%)")
```
